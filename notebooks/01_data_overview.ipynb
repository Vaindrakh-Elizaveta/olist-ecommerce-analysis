## Первичные наблюдения

- Таблица не содержит отдельного уникального идентификатора строки.
- Столбец `geolocation_zip_code_prefix` не является первичным ключом, поскольку один почтовый индекс может встречаться в нескольких записях.
- Каждая строка соответствует одной географической точке с координатами, городом и штатом.
- В таблице обнаружено 261 831 полностью дублирующаяся строка.
- Для таблицы `geolocation` это не обязательно свидетельствует об ошибке данных, так как одна и та же географическая точка может встречаться несколько раз.
- Решение об удалении дубликатов будет принято на этапе очистки данных после определения роли этой таблицы в анализе.

### Проверка составного ключа

## Первичные наблюдения

- Таблица содержит информацию о товарах, входящих в состав заказов.
- Каждая строка соответствует одному товару в конкретном заказе.
- Столбец `order_id` не является уникальным, поскольку один заказ может содержать несколько товаров.
- Столбец `order_item_id` также не является уникальным сам по себе — он представляет собой порядковый номер товара внутри заказа.
- Уникальной является комбинация столбцов `order_id` и `order_item_id`, которая выступает составным ключом таблицы.
- На этапе очистки данных столбец `shipping_limit_date` потребуется преобразовать из строкового типа `object` в тип `datetime`.
- Таблица содержит больше строк, чем таблица `orders`, что говорит о наличии заказов, включающих несколько товаров.

- Это свидетельствует о систематическом характере пропусков, а не о случайной потере отдельных значений.
