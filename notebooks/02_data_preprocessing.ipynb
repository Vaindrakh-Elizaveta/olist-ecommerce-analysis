```python
import pandas as pd
import numpy as np
```

```python
orders = pd.read_csv("../data/raw/olist_orders_dataset.csv")
order_items = pd.read_csv("../data/raw/olist_order_items_dataset.csv")
order_reviews = pd.read_csv("../data/raw/olist_order_reviews_dataset.csv")
order_payments = pd.read_csv("../data/raw/olist_order_payments_dataset.csv")
products = pd.read_csv("../data/raw/olist_products_dataset.csv")
product_category_name_translation = pd.read_csv("../data/raw/product_category_name_translation.csv")
customers = pd.read_csv("../data/raw/olist_customers_dataset.csv")
sellers = pd.read_csv("../data/raw/olist_sellers_dataset.csv")
geolocation = pd.read_csv("../data/raw/olist_geolocation_dataset.csv")
```

### Преобразование дат

```python
orders["order_purchase_timestamp"] = pd.to_datetime(orders["order_purchase_timestamp"])
orders["order_approved_at"] = pd.to_datetime(orders["order_approved_at"])
orders["order_delivered_carrier_date"] = pd.to_datetime(orders["order_delivered_carrier_date"])
orders["order_delivered_customer_date"] = pd.to_datetime(orders["order_delivered_customer_date"])
orders["order_estimated_delivery_date"] = pd.to_datetime(orders["order_estimated_delivery_date"])

order_items["shipping_limit_date"] = pd.to_datetime(order_items["shipping_limit_date"])
order_reviews["review_creation_date"] = pd.to_datetime(order_reviews["review_creation_date"])
order_reviews["review_answer_timestamp"] = pd.to_datetime(order_reviews["review_answer_timestamp"])
```

### Проверка дат на логичность

На данном этапе проверяются даты из таблицы `orders` на логичную последовательность: покупка, подтверждение, передача перевозчику, доставка клиенту.

```python
orders[orders["order_purchase_timestamp"] > orders["order_approved_at"]]
```

```python
orders[orders["order_approved_at"] > orders["order_delivered_carrier_date"]]
```

```python
delta = (
    orders["order_approved_at"] - orders["order_delivered_carrier_date"]
)
delta[delta > pd.Timedelta(0)].describe()
```

```python
delta.sort_values(ascending=False).head(10)
```

```python
orders.loc[
    delta.sort_values(ascending=False).head(10).index,
    [
        "order_purchase_timestamp",
        "order_approved_at",
        "order_delivered_carrier_date",
        "order_delivered_customer_date"
    ]
]
```

Проверка логической последовательности дат выявила 1359 заказов, у которых дата подтверждения заказа `order_approved_at` позже даты передачи перевозчику    order_delivered_carrier_date    . Для большинства записей нарушение составляет менее суток, однако встречаются отдельные экстремальные случаи `до 171 дня`, которые с высокой вероятностью являются ошибками исходных данных. На данном этапе изменения в таблицу не вносятся. Решение об обработке таких записей будет принято после завершения исследовательского анализа и оценки их влияния на результаты.

```python
orders[orders["order_delivered_carrier_date"] > orders["order_delivered_customer_date"]]
```

```python
orders[orders["order_delivered_carrier_date"] > orders["order_delivered_customer_date"]].shape
```

```python
delta = (
    orders["order_delivered_carrier_date"] - orders["order_delivered_customer_date"]
)

delta.sort_values(ascending=False).head(10)
```

При проверке последовательности этапов доставки обнаружено 23 заказа `≈0,02% от общего числа`, у которых дата передачи заказа перевозчику `order_delivered_carrier_date` позже даты доставки клиенту `order_delivered_customer_date`. Такие записи противоречат ожидаемой логике процесса доставки и, вероятнее всего, являются ошибками исходных данных. Поскольку доля подобных случаев крайне мала и невозможно однозначно определить некорректное значение, на данном этапе записи сохраняются без изменений.

Такие записи противоречат ожидаемой логике данных и, вероятно, связаны с особенностями формирования исходной базы данных или редкими ошибками. Поскольку их количество составляет менее 0.01% от общего объёма данных, на данном этапе они сохраняются без изменений.
